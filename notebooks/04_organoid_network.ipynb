# Organoid network

## Scientific objective

Build and inspect the study-provenanced heterogeneous evidence network.

## Background and data sources

The input is the identifier-verified curated CSV. Primary records are linked through PubMed. See `docs/search_strategy.md` and `docs/data_provenance.md`.


## Methods

Reusable functions from `src/` are used so notebook and command-line results share one implementation.


```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from src.data_extraction import load_studies
studies = load_studies(ROOT / 'data' / 'processed' / 'organoid_study_database.csv')
print(f'Loaded {len(studies)} curated studies')
```

```python
from src.network_analysis import build_network, network_metrics
graph = build_network(studies)
display(network_metrics(graph))
print('All edges have provenance:', all(d.get('study_ids') for *_, d in graph.edges(data=True)))
```

## Biological interpretation

Network structure describes documented relationships and cannot establish therapeutic readiness.


## Limitations

Single-reviewer bounded evidence map; heterogeneous models and endpoints; incomplete safety reporting; missing evidence is not a negative result; no causal or clinical-readiness inference from counts.
